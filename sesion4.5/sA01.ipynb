Solicitar un precio y determinar el precio más IVA

```python
precio = float(input("Dame el precio:")) # Recupera un texto del usuario

iva = precio * 0.16
precio_mas_iva = precio + iva

print(f"Precio original: $ {precio:.2f}")
print(f"Precio más iva:  $ {precio_mas_iva:.2f}")
```

Datos de un ticket con formato

```python
producto1 = "Coca Cola 600ml"
precio1 = 24.5
cantidad1 = 5
total1 = precio1 * cantidad1

producto2 = "Gansito"
precio2 = 28.5
cantidad2 = 2
total2 = precio2 * cantidad2

producto3 = "Galletas marías"
precio3 = 16.90
cantidad3 = 7
total3 = precio3 * cantidad3

ANCHO_TICKET = 52

nombre_tienda = "Abarrotes Don Pancho"
fecha = "26/09/2026"
print(f"{nombre_tienda:^52s}")
print(f"{fecha:^52s}")
print("-" * ANCHO_TICKET)
print(f"{producto1:20s} $ {precio1:8.2f}   x {cantidad1:02d}    $ {total1:8.2f}")
print(f"{producto2:20s} $ {precio2:8.2f}   x {cantidad2:02d}    $ {total2:8.2f}")
print(f"{producto3:20s} $ {precio3:8.2f}   x {cantidad3:02d}    $ {total3:8.2f}")
print("-" * ANCHO_TICKET)
print(" " * (ANCHO_TICKET - 10) + f"$ {total1 + total2 + total3:8.2f}")
```
