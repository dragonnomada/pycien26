# Proyecto 1 - Sistema de Administración de Venta de Frutas

Una tienda tiene una lista de frutas y una lista de precios, en un menú el sistema puede registrar una fruta, eliminar una fruta, modificar un precio y determinar vender una fruta con la cantidad aplicable (el precio es por kilo, entonces, las ventas son por kilo).

1. Cuando se crea una fruta se debe insertar el precio para esa fruta en cero
2. Cuando se elimina una fruta se debe quitar su precio de lista de precios también
3. Cuando se modifica el precio de una fruta solo se reemplaza el valor del precio

Este problema es especialmente difícil, porque tenemos que mantener dos listas sincronizadas, para que los índices de una lista coincidan con los elementos de la otra lista, por ejemplo, la lista de frutas debe tener su precio en la misma posición siempre, si se agrega o quita una fruta se hará sobrea ambas listas, y si se desea consultar el precio de una fruta, primero se deberá buscar el índice sobre la lista de frutas y luego su precio en la otra lista, pero con el mismo índice.

El menú se muestra siempre y él se muestran las frutas y sus precios previamiente, al estilo cartelera (usar formato avanzado).

* **Nota:** No se deben poder insertar dos frutas con mismo nombre

```python
def imprimirListaPrecios(frutas, precios):
    print("Lista de precios:")
    print("-" * 40) # Línea de 40-guiones

    for indice in range(0, len(frutas)):
        fruta = frutas[indice]
        precio = precios[indice]

        print(f"{indice + 1}. {fruta:20s} $ {precio:6.2f}")

    print("-" * 40) # Línea de 40-guiones
```

```python
def imprimirMenu():
    # Mostrar el menú
    print("Menú de opciones:")
    print("-" * 40) # Línea de 40-guiones
    print("1. Agregar una fruta con precio $ 0.00")
    print("2. Modificar el precio de una fruta")
    print("3. Eliminar una fruta y su precio")
    print("-" * 40) # Línea de 40-guiones
    print("4. Calcular el total venta")
    print("   para una fruta y cantidad")
    print()
```

```python
def seleccionarOpcion():
    opcion = input("Selecciona la opción del menú: ")
    print(f"Opcion seleccionada: {opcion}")
    try:
        return int(opcion)
    except:
        return None
```

```python
def agregarFruta(frutas, precios):
    nombre = input("Dame el nombre de la fruta:")

    if nombre in frutas:
        print(f"LA FRUTA {nombre} YA EXISTE")
        return
    else:
        frutas.append(nombre)
        precios.append(0)
        print(f"FRUTA AGREGADA {nombre} con precio $ 0.00")
```

```python
def modificarPrecioFruta(frutas, precios):
    nombre = input("Dame el nombre de la fruta:")
    
    if not nombre in frutas:
        print(f"ERROR: NO EXISTE LA FRUTA {nombre}")
        return
    else:
        precio_nuevo = float(input("Dame el precio nuevo:"))
        indice = frutas.index(nombre)
        precios[indice] = precio_nuevo
        print(f"LA FRUTA {nombre} HA CAMBIADO SU PRECIO A $ {precio_nuevo:.2f}")
```

```python
frutas = ["manzana", "pera", "mango"]
precios = [100, 40, 150]
```

```python
imprimirListaPrecios(frutas, precios)

imprimirMenu()

opcion = seleccionarOpcion()

if opcion == 1:
    print("AGREGAR FRUTA CON PRECIO $ 0.00", end="\n\n")
    agregarFruta(frutas, precios)
elif opcion == 2:
    print("MODIFICAR PRECIO DE UNA FRUTA", end="\n\n")
    modificarPrecioFruta(frutas, precios)
elif opcion == 3:
    print("ELIMINAR UNA FRUTA Y SU PRECIO", end="\n\n")
elif opcion == 4:
    print("CALCULAR UNA VENTA", end="\n\n")
else:
    print("OPCIÓN NO VÁLIDA")
```
