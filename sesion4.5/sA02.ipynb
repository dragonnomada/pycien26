Calcular el precio más IVA

```python
def calcularPrecioMasIVA(precio):
    iva = precio * 0.16
    return precio + iva

calcularPrecioMasIVA(1000)
```

```python
calcularPrecioMasIVA(2000)
```

```python
calcularPrecioMasIVA(24.5)
```

```python
def calculaAreaTriangulo(l1, l2, l3):
    s = (l1 + l2 + l3) / 2
    discriminante = s * (s - l1) * (s - l2) * (s - l3)
    if discriminante < 0:
        raise(Exception("El área es negativa, el triángulo no es válido"))
    area = discriminante ** 0.5
    return area
```

```python
calculaAreaTriangulo(1, 1, 1)
```

```python
calculaAreaTriangulo(2, 2, 2)
```
