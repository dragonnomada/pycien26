```python
titanic["Cabin"].isna() # Serie lógica de valores nulos (no tener cabina)
```

```python
e1 = titanic["Survived"] # Binaria (Numérico / Categórico)
e2 = titanic["Pclass"] # 3-clases (1, 2, 3) (Numérico / Categórico)
e3 = titanic["Sex"] # Binaria (Categórica)
e4 = titanic["Age"] # Numérico
e5 = titanic["SibSp"] + titanic["Parch"] # Numérico
e6 = titanic["Fare"] # Numérico
e7 = ~titanic["Cabin"].isna() # Binaria (Categórico) # ~ negar una serie, .isna() si está vació
e8 = titanic["Embarked"] # 3-clases (S, C, Q) (Categórico)
```

```python
e1.value_counts()
```
