> Definición de la clase

```python
class Persona:

    nombre: str = None
    edad: str = None
    salario: str = None

    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        self.salario = 0.0

    def describir(self):
        print(f"💁 {self.nombre}, {self.edad} años")
        print("-" * 80)
        print(f"💸 $ {self.salario:.2f}")

    # [MODIFICADO]
    def indiceSalarial(self):
        return self.salario / self.edad
```

> Inicialización de los objetos de la clase (`instancias`)

```python
persona1 = Persona("Any", 26)
persona2 = Persona("Beny", 32)
```

> El objeto `persona1` es una instancia de la clase Persona, con una dirección de memoria, mediante la cuál se retienen todos sus atributos, dentro de la clase este es el `self` o la referencia al objeto

```python
persona1
```

> A través del objeto `persona1`, hacemos `persona1.nombre` para recuperar el valor retenido por la clase en ese atributo, `persona1.edad` recuperará la edad y `persona1.salario` recuperará el salario

```python
persona1.nombre
```

```python
persona1.edad
```

```python
persona1.salario
```

> Los métodos dentro de la clase, pueden ser invocados y `self` será inferido como el objeto de la instancia que lo invoca, en este caso, `persona1` asume el rol de `self` y por lo tanto no se pasa como parámetro, solo los demás parámetros si es que hay

```python
persona1.describir()
```

> Los objetos permiten alterar el valor de atributos en cualquier momento, por lo que al invocar las funciones (métodos), estos asumirán los últimos valores retenidos para realizar sus cálculos

```python
persona1.salario = 12500

persona1.describir()
```

```python
persona2 = Persona("Beny", 32)
persona2.salario = 15800

persona2.describir()
```

```python
persona3 = Persona("Albertano", 45)
persona3.salario = 4000

persona3.describir()
```

```python
persona1.indiceSalarial()
```

```python
persona2.indiceSalarial()
```

```python
persona3.indiceSalarial()
```
