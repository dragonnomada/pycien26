```python
import numpy
import pandas

from matplotlib import pyplot
import seaborn

import keras
```

```python
students = pandas.read_csv("../conjuntos/student_habits_performance.csv")

students
```

```python
x1 = students["age"]
x2 = students["gender"] == "Female"
x3 = students["gender"] == "Other"
x4 = students["study_hours_per_day"]
x5 = students["social_media_hours"]
x6 = students["netflix_hours"]
x7 = students["part_time_job"] == "Yes"

y1 = students["exam_score"]

X = pandas.DataFrame({
    "x1": x1,
    "x2": x2,
    "x3": x3,
    "x4": x4,
    "x5": x5,
    "x6": x6,
    "x7": x7,
}, dtype="float32")

Y = pandas.DataFrame({
    "y1": y1
}, dtype="float32")

X
```

```python
Y
```

```python
modelo = keras.Sequential([
    keras.Input(shape=(7,)), # 7 - características
    # keras.layers.Dense(16, activation="relu"),
    # keras.layers.Dense(8, activation="relu"), # Oculta: relu
    keras.layers.Dense(1, activation="linear"), # Abierta: linear | Cerrada: sigmoid
])

modelo.compile(
    optimizer="adam", # Stochastic Gradient Descend
    loss="mse", # Abierta: mse | Cerrada: binary_crossentropy
    metrics=["mae"] # Abierta: mae | Cerrada: accuracy
)

modelo.summary()
```

```python
modelo.fit(X, Y, epochs=1000)
```

```python
modelo.save("students_1x1_linear.keras")
```
